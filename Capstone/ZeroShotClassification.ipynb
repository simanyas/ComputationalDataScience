```python
import re
def clean_tweet(tweet):
    return ' '.join(re.sub('(\\\\n)|(b\"[^0-9A-Za-z A-Za-z0-9 \t]+)|(b\'[^0-9A-Za-z]+)|(b\"[A-Za-z0-9]+)|(b\'[A-Za-z0-9]+)|(b\'#[A-Za-z0-9]+)|(b\'@[A-Za-z0-9]+)|(\\\\x[A-Za-z0-9]+)|(@[A-Za-z0-9]+)|([^0-9A-Za-z \t])|(\w+:\/\/\S+)|([RT])', ' ', str(tweet).lower()).split())
```

```python
# Call function to get Clean tweets
df["CleanTweet"] = df['text'].apply(lambda x : clean_tweet(x))
```

```python
df.drop(df.index[df.CleanTweet.eq("")], inplace=True)
```

```python
from flair.models import TARSClassifier
from flair.data import Sentence

# 1. Load our pre-trained TARS model for English
tars = TARSClassifier.load('tars-base')

# 3. Define some classes that you want to predict using descriptive names
classes = ["happy","sad","angry","fear","surprise","disgust","hate","worry"]
```

```python
def pred_zeroshot(sentence):
  tars.predict_zero_shot(sentence, classes,multi_label=False)
  #print(sentence)
  return sentence
```

```python
df["ZeroShot"] = df['CleanTweet'].apply(lambda x : pred_zeroshot(Sentence(x)))
```

```python
df.to_csv("ZeroShotDataset2022.csv",index=False)
```
